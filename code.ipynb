### **Uji Analisis Autokorelasi dan Normalitas Residual**

Untuk melakukan analisis ini, kita pertama-tama perlu memodelkan data (misalnya dengan regresi linear sederhana atau memisahkan tren waktu) untuk mendapatkan nilai **residual** (selisih antara nilai aktual dan nilai prediksi).

Kita akan melakukan analisis ini pada data **Padi** dan **Jagung** (menggunakan data non-null).

```python
import numpy as np
import pandas as pd
import statsmodels.api as sm
from statsmodels.stats.stattools import durbin_watson
from scipy import stats
import matplotlib.pyplot as plt

# Membaca data jika belum termuat di runtime ini
try:
    padi
    jagung
except NameError:
    padi = pd.read_excel('/content/Produksi_Padi_Nasional.xlsx')
    jagung = pd.read_excel('/content/Produksi_Jagung_Nasional.xlsx')

# Mempersiapkan data non-null untuk analisis
padi_clean = padi.dropna(subset=['Produksi_Ton']).copy()
jagung_clean = jagung.dropna(subset=['Produksi_Ton']).copy()

# Membuat variabel waktu numerik sebagai prediktor sederhana untuk mendapatkan residual
padi_clean['Waktu'] = np.arange(len(padi_clean))
jagung_clean['Waktu'] = np.arange(len(jagung_clean))

# --- 1. Pemodelan untuk Padi ---
X_padi = sm.add_constant(padi_clean['Waktu'])
y_padi = padi_clean['Produksi_Ton']
model_padi = sm.OLS(y_padi, X_padi).fit()
residual_padi = model_padi.resid

# --- 2. Pemodelan untuk Jagung ---
X_jagung = sm.add_constant(jagung_clean['Waktu'])
y_jagung = jagung_clean['Produksi_Ton']
model_jagung = sm.OLS(y_jagung, X_jagung).fit()
residual_jagung = model_jagung.resid

print("Residual berhasil dihitung untuk kedua dataset.")
```

#### **1. Uji Autokorelasi (Durbin-Watson Test)**

Nilai Durbin-Watson (DW) berkisar antara 0 hingga 4:
- Nilai mendekati **2** menunjukkan tidak ada autokorelasi.
- Nilai mendekati **0** menunjukkan autokorelasi positif.
- Nilai mendekati **4** menunjukkan autokorelasi negatif.

```python
dw_padi = durbin_watson(residual_padi)
dw_jagung = durbin_watson(residual_jagung)

print("=== HASIL UJI DURBIN-WATSON ===")
print(f"Durbin-Watson Padi   : {dw_padi:.4f}")
print(f"Durbin-Watson Jagung : {dw_jagung:.4f}")
```

#### **2. Uji Normalitas Residual (Shapiro-Wilk & Kolmogorov-Smirnov)**

Hipotesis:
- **H0**: Residual terdistribusi normal (p-value > 0.05)
- **H1**: Residual tidak terdistribusi normal (p-value <= 0.05)

```python
# Uji Shapiro-Wilk
shapiro_padi_stat, shapiro_padi_p = stats.shapiro(residual_padi)
shapiro_jagung_stat, shapiro_jagung_p = stats.shapiro(residual_jagung)

# Uji Kolmogorov-Smirnov (dibandingkan dengan distribusi normal standar yang disesuaikan)
ks_padi_stat, ks_padi_p = stats.kstest(residual_padi, 'norm', args=(residual_padi.mean(), residual_padi.std()))
ks_jagung_stat, ks_jagung_p = stats.kstest(residual_jagung, 'norm', args=(residual_jagung.mean(), residual_jagung.std()))

print("=== HASIL UJI NORMALITAS RESIDUAL ===")
print("\n1. PADI:")
print(f"- Shapiro-Wilk        : Stat={shapiro_padi_stat:.4f}, p-value={shapiro_padi_p:.4e}")
print(f"- Kolmogorov-Smirnov  : Stat={ks_padi_stat:.4f}, p-value={ks_padi_p:.4e}")

print("\n2. JAGUNG:")
print(f"- Shapiro-Wilk        : Stat={shapiro_jagung_stat:.4f}, p-value={shapiro_jagung_p:.4e}")
print(f"- Kolmogorov-Smirnov  : Stat={ks_jagung_stat:.4f}, p-value={ks_jagung_p:.4e}")
```

#### **3. Visualisasi Q-Q Plot Residual**

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 6))

# Q-Q Plot Padi
sm.qqplot(residual_padi, line='s', ax=axes[0])
axes[0].set_title("Q-Q Plot Residual Produksi Padi")

# Q-Q Plot Jagung
sm.qqplot(residual_jagung, line='s', ax=axes[1])
axes[1].set_title("Q-Q Plot Residual Produksi Jagung")

plt.tight_layout()
plt.show()
```

#### **4. Analisis Tambahan: Plot ACF dan PACF Residual**

Untuk mengidentifikasi struktur lag pada autokorelasi secara lebih detail, kita akan membuat grafik ACF (Autocorrelation Function) dan PACF (Partial Autocorrelation Function) dari residual model padi dan jagung.

```python
import statsmodels.api as sm
import matplotlib.pyplot as plt

fig, axes = plt.subplots(2, 2, figsize=(16, 10))

# --- Padi ---
# ACF Padi
sm.graphics.tsa.plot_acf(residual_padi, lags=20, ax=axes[0, 0], title="ACF Residual Produksi Padi")
# PACF Padi
sm.graphics.tsa.plot_pacf(residual_padi, lags=20, ax=axes[0, 1], title="PACF Residual Produksi Padi", method='yw')

# --- Jagung ---
# ACF Jagung
sm.graphics.tsa.plot_acf(residual_jagung, lags=20, ax=axes[1, 0], title="ACF Residual Produksi Jagung")
# PACF Jagung
sm.graphics.tsa.plot_pacf(residual_jagung, lags=20, ax=axes[1, 1], title="PACF Residual Produksi Jagung", method='yw')

plt.tight_layout()
plt.show()
```

berdasarkan plot acf pacf diatas, pada produksi padi, terdapat banyak lag yang keluar secara signifikan dari pita kepercayaan. ini beberarti bahawa tidak stasioner dana mengandung pola musiman. Kemudian Data perlu dilakukan proses differencing musiman (misal $d=1$ dan/atau $D=1$) agar data menjadi stasioner sebelum memodelkan menggunakan SARIMA.

### **Analisis Kebutuhan Transformasi Data**

Berdasarkan hasil analisis uji asumsi sebelumnya, berikut pertimbangan perlunya transformasi:

1. **Transformasi Stasioneritas (Time Series):**
   * **Padi:** Plot ACF/PACF menunjukkan pola musiman kuat (*seasonal*) dan penurunan lambat yang menandakan data **tidak stasioner**. Model time-series seperti ARIMA/SARIMA mewajibkan data stasioner. Maka, diperlukan **Transformasi Differencing (Pembedaan)**, baik non-musiman ($d=1$) maupun musiman ($D=1$).
   * **Jagung:** Meskipun ACF menunjukkan pola yang lebih stabil, kita perlu menguji stasioneritasnya secara formal.

2. **Transformasi Normalitas & Stabilisasi Varians:**
   * **Padi:** Residualnya terbukti tidak normal (Shapiro-Wilk $p < 0.05$). Transformasi seperti **Logaritma (Log Transform)** atau **Box-Cox** dapat membantu menstabilkan varians dan membuat residual lebih mendekati distribusi normal.
   * **Jagung:** Residual sudah terdistribusi normal ($p \approx 0.90$), sehingga tidak memerlukan transformasi normalitas.

Mari kita lakukan uji formal **Augmented Dickey-Fuller (ADF)** untuk membuktikan stasioneritas data asli.

```python
from statsmodels.tsa.stattools import adfuller

def uji_adf(series, nama_data):
    hasil = adfuller(series)
    print(f"=== UJI ADF: {nama_data.upper()} ===")
    print(f"ADF Statistic: {hasil[0]:.4f}")
    print(f"p-value: {hasil[1]:.4f}")
    print("Critical Values:")
    for key, value in hasil[4].items():
        print(f"   {key}: {value:.4f}")
    if hasil[1] < 0.05:
        print("Kesimpulan: Data STASIONER (Tolak H0)\n")
    else:
        print("Kesimpulan: Data TIDAK STASIONER (Gagal Tolak H0), membutuhkan Differencing.\n")

uji_adf(padi['Produksi_Ton'], "Produksi Padi")
uji_adf(jagung['Produksi_Ton'], "Produksi Jagung")
```

### **Penerapan Transformasi Differencing**

Karena hasil uji ADF menunjukkan kedua data tidak stasioner, kita akan menerapkan **Differencing tingkat 1 ($d=1$)** dan menguji ulang stasioneritasnya.

```python
# Melakukan Differencing tingkat 1
padi_diff = padi['Produksi_Ton'].diff().dropna()
jagung_diff = jagung['Produksi_Ton'].diff().dropna()

# Uji ulang dengan ADF
print("=== SETELAH DIFFERENCING TINGKAT 1 ===\n")
uji_adf(padi_diff, "Produksi Padi (Diff 1)")
uji_adf(jagung_diff, "Produksi Jagung (Diff 1)")

# Visualisasi tren setelah differencing
fig, axes = plt.subplots(2, 1, figsize=(14, 10))
axes[0].plot(padi_diff, color='green', marker='o')
axes[0].set_title('Produksi Padi setelah Differencing 1 (Stasioner)')
axes[0].grid(True)

axes[1].plot(jagung_diff, color='orange', marker='o')
axes[1].set_title('Produksi Jagung setelah Differencing 1 (Stasioner)')
axes[1].grid(True)

plt.tight_layout()
plt.show()
```

## **0. Setup, reproducibility, dan pembagian data latih-uji**

```python
# ============================================================
# SETUP, REPRODUCIBILITY, PEMBAGIAN LATIH-UJI
# ============================================================
import sys, warnings
import numpy as np
import pandas as pd
import scipy, statsmodels
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
from statsmodels.tsa.seasonal import STL
from statsmodels.tsa.stattools import adfuller, kpss, acf, pacf
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf
from statsmodels.tsa.holtwinters import ExponentialSmoothing
from statsmodels.tsa.statespace.sarimax import SARIMAX
from statsmodels.stats.diagnostic import acorr_ljungbox, het_arch
from statsmodels.stats.stattools import jarque_bera

# Peringatan numerik dari statsmodels disembunyikan agar keluaran rapi.
# Konvergensi tiap model tetap diperiksa eksplisit (kolom "Konvergen").
warnings.filterwarnings('ignore')

SEED = 42            # seed untuk simulasi interval (Exponential Smoothing)
np.random.seed(SEED)
M = 12               # periode musiman (data bulanan)
H_UJI = 12           # panjang data uji (12 bulan terakhir)
H_RAMAL = 12         # horizon peramalan ke depan
ALPHA = 0.05

print('Versi pustaka (catat untuk laporan):')
print(f'  python      {sys.version.split()[0]}')
print(f'  numpy       {np.__version__}')
print(f'  pandas      {pd.__version__}')
print(f'  scipy       {scipy.__version__}')
print(f'  statsmodels {statsmodels.__version__}')
print(f'  seed        {SEED}')

# Deret bulanan kontinu (setelah baris kosong dihapus pada bagian PREPROCESING)
SERI = {}
for nama, df in [('Padi', padi), ('Jagung', jagung)]:
    s = df.set_index('Tanggal')['Produksi_Ton'].astype(float).sort_index().asfreq('MS')
    assert s.isna().sum() == 0, f'{nama}: masih ada data hilang di tengah deret'
    s.name = nama
    SERI[nama] = s

# Pembagian latih-uji: 12 bulan terakhir sebagai data uji
TRAIN = {n: s.iloc[:-H_UJI] for n, s in SERI.items()}
TEST  = {n: s.iloc[-H_UJI:] for n, s in SERI.items()}

ringkas = pd.DataFrame({
    n: {
        'Periode data': f'{s.index[0]:%Y-%m} s.d. {s.index[-1]:%Y-%m}',
        'Jumlah titik': len(s),
        'Data latih': f'{len(TRAIN[n])} ({TRAIN[n].index[0]:%Y-%m} s.d. {TRAIN[n].index[-1]:%Y-%m})',
        'Data uji': f'{len(TEST[n])} ({TEST[n].index[0]:%Y-%m} s.d. {TEST[n].index[-1]:%Y-%m})',
    } for n, s in SERI.items()
})
print()
print(ringkas.to_string())

# Berkas data (.csv) yang dipakai analisis, sesuai ketentuan pengumpulan
for n, s in SERI.items():
    s.to_csv(f'data_{n.lower()}_bulanan.csv', header=['Produksi_Ton'])

# Wadah hasil model (diisi pada tahap-tahap berikutnya)
HASIL = {n: {} for n in SERI}
```

## **TAHAP 1. Eksplorasi data (lanjutan)**

Plot deret, boxplot per tahun, dan seasonal plot sudah ada di bagian EDA di atas. Berikut statistik ringkas, deteksi pencilan, dekomposisi, boxplot per bulan, dan ACF/PACF data mentah, seluruhnya pada **data latih**.

```python
print('=== STATISTIK RINGKAS (data latih, ton) ===')
print(pd.DataFrame({n: t.describe() for n, t in TRAIN.items()}).round(0).to_string())

print('\n=== PENCILAN: residual STL robust, ambang 3 sigma-robust (data latih) ===')
for n, y in TRAIN.items():
    r = STL(y, period=M, robust=True).fit().resid
    sigma = 1.4826 * np.median(np.abs(r - np.median(r)))
    out = r[np.abs(r - np.median(r)) > 3 * sigma]
    print(f'{n}: {len(out)} titik terdeteksi')
    for t, v in out.items():
        print(f'   {t:%Y-%m}  nilai = {y.loc[t]:,.0f}  residual STL = {v:,.0f}')
print('\nPencilan yang terdeteksi TIDAK dihapus. Titik tersebut dipertahankan karena merupakan produksi riil, dan keberadaannya dibahas pada diagnostik residual.')
```

```python
# ---- Dekomposisi (STL robust) dan pemilihan aditif vs multiplikatif ----
# Kriteria: deret dikerjakan secara multiplikatif bila dekomposisi pada skala log
# memberi sisa relatif yang lebih kecil daripada dekomposisi aditif.
PAKAI_LOG, KUAT = {}, {}
baris = []
for n, y in TRAIN.items():
    st_add = STL(y, period=M, robust=True).fit()
    st_log = STL(np.log(y), period=M, robust=True).fit()
    galat_add = np.std(st_add.resid / (st_add.trend + st_add.seasonal))   # sisa relatif (aditif)
    galat_mul = np.std(st_log.resid)                                      # sisa relatif (multiplikatif)

    lengkap = y.groupby(y.index.year).filter(lambda g: len(g) == M)       # tahun penuh saja
    th = lengkap.groupby(lengkap.index.year).agg(['mean', 'std'])
    rho, p_rho = stats.spearmanr(th['mean'], th['std'])

    PAKAI_LOG[n] = bool(galat_mul < galat_add)
    st = st_log if PAKAI_LOG[n] else st_add
    Fs = max(0.0, 1 - np.var(st.resid) / np.var(st.resid + st.seasonal))
    Ft = max(0.0, 1 - np.var(st.resid) / np.var(st.resid + st.trend))
    KUAT[n] = dict(Fs=Fs, Ft=Ft, stl=st)
    baris.append({'Deret': n, 'Sisa relatif aditif': galat_add, 'Sisa relatif multiplikatif': galat_mul,
                  'Keputusan': 'multiplikatif (log)' if PAKAI_LOG[n] else 'aditif',
                  'Spearman rerata-vs-sd tahunan': rho, 'p': p_rho, 'n tahun': len(th),
                  'Kekuatan tren Ft': Ft, 'Kekuatan musiman Fs': Fs})
tabel_dek = pd.DataFrame(baris).set_index('Deret')
print(tabel_dek.round(3).T.to_string())
print('\nCatatan: korelasi rerata-vs-sd tahunan dihitung dari sedikit tahun penuh, jadi hanya pendukung. Keputusan utama memakai perbandingan sisa relatif.')

for n in SERI:
    st = KUAT[n]['stl']
    skala = 'log' if PAKAI_LOG[n] else 'asli (aditif)'
    fig, ax = plt.subplots(4, 1, figsize=(13, 9), sharex=True)
    for a, (nm, v) in zip(ax, [('Observasi', st.observed), ('Tren', st.trend),
                               ('Musiman', st.seasonal), ('Residual', st.resid)]):
        a.plot(v, color='tab:green' if n == 'Padi' else 'tab:orange')
        a.set_ylabel(nm)
        a.grid(True, alpha=0.4)
    ax[0].set_title(f'Dekomposisi STL {n} (data latih, skala {skala})  Ft={KUAT[n]["Ft"]:.2f}, Fs={KUAT[n]["Fs"]:.2f}', fontweight='bold')
    ax[-1].set_xlabel('Tanggal')
    plt.tight_layout()
    plt.show()
```

## **TAHAP 2. Transformasi dan stasioneritas**

Urutan keputusan: (1) transformasi varians (log atau tidak), (2) uji ADF dan KPSS pada deret asli, setelah differencing biasa, dan setelah differencing musiman, (3) penetapan *d* dan *D* dengan memilih differencing **paling sedikit** yang membuat deret stasioner (ADF menolak H0 akar unit **dan** KPSS gagal menolak H0 stasioner), (4) pemeriksaan *over-differencing*.

```python
# ---- Transformasi varians ----
print('=== BOX-COX (lambda diestimasi dari DATA LATIH) ===')
for n, y in TRAIN.items():
    _, lam, (lo, hi) = stats.boxcox(y.values, alpha=0.05)
    ket = []
    if lo <= 0 <= hi: ket.append('CI 95% memuat 0 -> log wajar')
    if lo <= 1 <= hi: ket.append('CI 95% memuat 1 -> tanpa transformasi wajar')
    print(f'{n}: lambda = {lam:.3f}, CI 95% = [{lo:.3f}, {hi:.3f}]  {"; ".join(ket) if ket else ""}')

print('\n=== KEPUTUSAN TRANSFORMASI (dari dekomposisi di Tahap 1) ===')
for n in SERI:
    print(f'{n}: ' + ('log (pola musiman proporsional terhadap level)' if PAKAI_LOG[n]
                      else 'tanpa log (amplitudo musiman relatif konstan); nilai dibagi 1e6 (juta ton) hanya agar optimisasi numerik stabil'))

def fwd(n, x):
    # transformasi maju
    return np.log(x) if PAKAI_LOG[n] else x / 1e6

def inv(n, x):
    # transformasi balik ke ton
    return np.exp(x) if PAKAI_LOG[n] else x * 1e6

YT = {n: fwd(n, TRAIN[n]) for n in SERI}      # deret latih pada skala model
```

```python
# ---- Uji stasioneritas: ADF (H0: akar unit) dan KPSS (H0: stasioner) ----
def beda(x, d, D, m=M):
    z = pd.Series(x).copy()
    for _ in range(D):
        z = z.diff(m)
    for _ in range(d):
        z = z.diff()
    return z.dropna()

def uji_stasioner(x):
    x = pd.Series(x).dropna()
    a = adfuller(x, autolag='AIC')
    try:
        k = kpss(x, regression='c', nlags='auto')
        k_stat, k_p = k[0], k[1]
    except Exception:
        k_stat, k_p = np.nan, np.nan
    return a[0], a[1], k_stat, k_p

baris = []
for n in SERI:
    y = YT[n]
    for D in (0, 1):
        for d in (0, 1, 2):
            if D == 1 and d == 2:
                continue
            z = beda(y, d, D)
            a_s, a_p, k_s, k_p = uji_stasioner(z)
            if a_p < ALPHA and k_p >= ALPHA: kes = 'stasioner (ADF & KPSS sepakat)'
            elif a_p < ALPHA:                kes = 'ADF stasioner, KPSS menolak'
            elif k_p >= ALPHA:               kes = 'ADF tidak menolak akar unit, KPSS tidak menolak'
            else:                            kes = 'tidak stasioner'
            baris.append({'Deret': n, 'd': d, 'D': D, 'n': len(z), 'Var': z.var(),
                          'ADF stat': a_s, 'ADF p': a_p, 'KPSS stat': k_s, 'KPSS p': k_p, 'Kesimpulan': kes})
TAB_STAS = pd.DataFrame(baris)
print(TAB_STAS.round(4).to_string(index=False))
print('\nCatatan: p-value KPSS pada statsmodels dibatasi pada rentang [0,01; 0,10] (interpolasi tabel), jadi 0,10 berarti "p >= 0,10".')
```

```python
# ---- Penetapan d dan D ----
def pilih_d(tab, D):
    t = tab[tab['D'] == D].sort_values('d')
    ok = t[(t['ADF p'] < ALPHA) & (t['KPSS p'] >= ALPHA)]
    if len(ok):
        return int(ok['d'].iloc[0]), 'ADF dan KPSS sepakat stasioner'
    ok = t[t['ADF p'] < ALPHA]
    if len(ok):
        return int(ok['d'].iloc[0]), 'hanya ADF yang menyatakan stasioner (KPSS masih menolak); d terkecil menurut ADF dipilih'
    return int(t['d'].max()), 'belum ada yang lolos; d terbesar yang diuji dipakai (perlu ditinjau)'

ORDE = {}
for n in SERI:
    tab = TAB_STAS[TAB_STAS['Deret'] == n]
    d_reg, alasan_reg = pilih_d(tab, 0)                      # untuk keluarga ARIMA tanpa musiman (D = 0)
    Fs = KUAT[n]['Fs']
    D = 1 if Fs > 0.64 else 0                                # differencing musiman bila musiman kuat
    d_sar, alasan_sar = pilih_d(tab, D)                      # untuk SARIMA
    ORDE[n] = dict(d_reg=d_reg, D=D, d_sar=d_sar)
    print(f'--- {n} ---')
    print(f'  ARIMA non-musiman : d = {d_reg}, D = 0   ({alasan_reg})')
    print(f'  Kekuatan musiman Fs = {Fs:.2f} -> D = {D} ' + ('(musiman kuat, differencing musiman diperlukan)' if D else '(musiman tidak cukup kuat untuk differencing musiman)'))
    print(f'  SARIMA            : d = {d_sar}, D = {D}   ({alasan_sar})')

# ---- Pemeriksaan over-differencing ----
print('\n=== PEMERIKSAAN OVER-DIFFERENCING ===')
print('Tanda over-differencing: varians naik setelah differencing tambahan dan/atau ACF lag-1 sangat negatif (< -0,5).')
for n in SERI:
    for (d, D) in sorted({(ORDE[n]['d_reg'], 0), (ORDE[n]['d_sar'], ORDE[n]['D'])}):
        z = beda(YT[n], d, D)
        r1 = acf(z, nlags=1, fft=False)[1]
        v_sebelum = [beda(YT[n], dd, D).var() for dd in range(0, d + 1)]
        naik = any(v_sebelum[i + 1] > v_sebelum[i] for i in range(len(v_sebelum) - 1))
        print(f'{n} (d={d}, D={D}): ACF lag-1 = {r1:.2f}; varians naik saat menambah d: {"ya" if naik else "tidak"}; '
              + ('WASPADA over-differencing' if (r1 < -0.5) else 'tidak ada indikasi kuat over-differencing'))

fig, axes = plt.subplots(2, 2, figsize=(15, 8))
for i, n in enumerate(SERI):
    for j, (d, D, judul) in enumerate([(ORDE[n]['d_reg'], 0, 'diff biasa'), (ORDE[n]['d_sar'], ORDE[n]['D'], 'diff biasa + musiman')]):
        z = beda(YT[n], d, D)
        axes[i, j].plot(z, color='tab:green' if n == 'Padi' else 'tab:orange', marker='o', ms=3)
        axes[i, j].axhline(0, color='k', lw=0.8)
        axes[i, j].set_title(f'{n}: d={d}, D={D} ({judul})', fontweight='bold')
        axes[i, j].grid(True, alpha=0.4)
plt.tight_layout()
plt.show()
```

**Tafsir singkat Tahap 2.** Hasil ADF dan KPSS dibaca bersama: stasioner bila ADF menolak H0 (p < 0,05) **dan** KPSS tidak menolak H0 (p >= 0,05). Bila keduanya tidak sepakat, hal itu dilaporkan apa adanya dan differencing terkecil menurut ADF dipakai. Differencing tidak ditambah hanya demi grafik tampak rata, dan kolom varians serta ACF lag-1 dipakai sebagai pemeriksa *over-differencing*. Untuk model non-musiman (AR, MA, ARMA, ARIMA) dipakai *d* dengan D = 0; untuk SARIMA dipakai pasangan (*d*, *D*) terpisah.

## **TAHAP 3. Moving Average dan Exponential Smoothing**

**3a. Moving Average (penghalus).** Jendela 3, 6, dan 12 dibandingkan. Untuk jendela genap dipakai rata-rata terpusat (jendela 12 menjadi MA 2x12) agar tidak bergeser setengah periode.

```python
JENDELA = [3, 6, 12]

def ma_terpusat(y, w):
    r = y.rolling(w, center=True).mean()
    return r if w % 2 == 1 else (r + r.shift(-1)) / 2

fig, axes = plt.subplots(2, 1, figsize=(14, 10))
baris = []
for ax, (n, y) in zip(axes, TRAIN.items()):
    ax.plot(y, color='lightgray', marker='o', ms=3, label='Observasi')
    for w in JENDELA:
        ma = ma_terpusat(y, w)
        ax.plot(ma, lw=2, label=f'MA({w})' + (' (2x12 terpusat)' if w == 12 else ''))
        baris.append({'Deret': n, 'Jendela': w,
                      'SD beda-pertama MA (kehalusan)': ma.diff().std(),
                      'Variasi musiman tersisa': (ma / ma.mean()).std()})
    ax.set_title(f'Moving Average pada Produksi {n} (data latih)', fontweight='bold')
    ax.set_ylabel('Produksi (ton)')
    ax.grid(True, alpha=0.4)
    ax.legend()
plt.tight_layout()
plt.show()
print(pd.DataFrame(baris).round(4).to_string(index=False))
```

Jendela 3 dan 6 masih menyisakan gelombang musiman (terlihat dari kolom "kehalusan" yang lebih besar). **3b. SES, Holt, dan Holt-Winters.** Delapan spesifikasi dicoba pada **data latih**. Pemilihan komponen tren dan musiman dalam satu keluarga memakai **AICc data latih**, bukan data uji. AIC antar-keluarga ES tidak dipakai untuk membandingkan dengan ARIMA.

```python
SPEK_ES = {
    'SES':                  dict(trend=None,  damped_trend=False, seasonal=None),
    'Holt':                 dict(trend='add', damped_trend=False, seasonal=None),
    'Holt-damped':          dict(trend='add', damped_trend=True,  seasonal=None),
    'HW aditif':            dict(trend='add', damped_trend=False, seasonal='add'),
    'HW aditif damped':     dict(trend='add', damped_trend=True,  seasonal='add'),
    'HW multiplikatif':     dict(trend='add', damped_trend=False, seasonal='mul'),
    'HW mult. damped':      dict(trend='add', damped_trend=True,  seasonal='mul'),
    'HW aditif tanpa tren': dict(trend=None,  damped_trend=False, seasonal='add'),
    'HW mult. tanpa tren':  dict(trend=None,  damped_trend=False, seasonal='mul'),
}
KELUARGA_ES = {
    'SES': ['SES'],
    'Holt': ['Holt', 'Holt-damped'],
    'Holt-Winters': ['HW aditif', 'HW aditif damped', 'HW multiplikatif', 'HW mult. damped',
                     'HW aditif tanpa tren', 'HW mult. tanpa tren'],
}

def fit_es(y, spek):
    mod = ExponentialSmoothing(y, trend=spek['trend'], damped_trend=spek['damped_trend'],
                               seasonal=spek['seasonal'], seasonal_periods=M if spek['seasonal'] else None,
                               initialization_method='estimated')
    return mod.fit()

def n_param_es(fit):
    p = fit.params
    return sum(1 for k in ('smoothing_level', 'smoothing_trend', 'smoothing_seasonal', 'damping_trend')
               if p.get(k) is not None and not (isinstance(p.get(k), float) and np.isnan(p.get(k))))

ES_SEMUA = {n: {} for n in SERI}
for n in SERI:
    y = TRAIN[n]                     # ES dikerjakan pada skala asli (ton)
    baris = []
    for nm, sp in SPEK_ES.items():
        try:
            f = fit_es(y, sp)
        except Exception as e:
            print(f'[gagal] {n} {nm}: {e}')
            continue
        ES_SEMUA[n][nm] = (sp, f)
        p = f.params
        baris.append({'Spesifikasi': nm, 'alpha (level)': p.get('smoothing_level'), 'beta (tren)': p.get('smoothing_trend'),
                      'gamma (musiman)': p.get('smoothing_seasonal'), 'phi (damping)': p.get('damping_trend'),
                      'level awal': p.get('initial_level'), 'tren awal': p.get('initial_trend'),
                      'SSE': f.sse, 'AICc': f.aicc})
    tab = pd.DataFrame(baris).set_index('Spesifikasi')
    print(f'=== {n}: parameter dan nilai awal Exponential Smoothing (data latih) ===')
    print(tab.round(4).to_string())
    print()

    for kel, daftar in KELUARGA_ES.items():
        ada = [d for d in daftar if d in ES_SEMUA[n]]
        if not ada:
            continue
        terbaik = min(ada, key=lambda d: ES_SEMUA[n][d][1].aicc)
        sp, f = ES_SEMUA[n][terbaik]
        HASIL[n][kel] = dict(kind='es', nama=terbaik, spek=sp, fit=f, k=n_param_es(f))
        tren = 'dengan tren' + (' (damped)' if sp['damped_trend'] else '') if sp['trend'] else 'tanpa tren'
        mus = {'add': 'musiman aditif', 'mul': 'musiman multiplikatif', None: 'tanpa musiman'}[sp['seasonal']]
        print(f'  Keluarga {kel:13s} -> {terbaik} ({tren}, {mus}); AICc terkecil di antara {ada}')
    print('-' * 100)
```

**Tafsir Tahap 3b.** Cetakan di atas menyatakan komponen mana yang dipilih per keluarga beserta alasannya (AICc data latih terkecil). Interpretasi parameter: *alpha* mendekati 1 berarti level bereaksi cepat terhadap observasi terbaru; *beta* kecil berarti tren berubah pelan; *gamma* menentukan seberapa cepat pola musiman diperbarui; *phi* < 1 meredam tren pada horizon panjang. Bila SES terpilih/hampir sama baiknya dengan Holt-Winters pada deret yang jelas musiman, itu tanda komponen musiman gagal terestimasi dan perlu dikritisi. Nilai awal level, tren, dan musiman diestimasi bersama parameter (`initialization_method='estimated'`).

## **TAHAP 4. Model AR, MA, dan ARMA**

Identifikasi orde memakai ACF/PACF deret **stasioner** hasil differencing biasa (*d* dari Tahap 2, D = 0). Aturan baca: PACF terpotong setelah lag *p* menunjuk AR(*p*), ACF terpotong setelah lag *q* menunjuk MA(*q*), keduanya meluruh menunjuk ARMA. Lag diperiksa pada 1 sampai 4 (lag musiman ditangani di SARIMA), batas signifikan = 1,96/sqrt(n). Dua kandidat diambil untuk tiap jenis.

Model diestimasi sebagai ARIMA(p, d, q) dengan *d* tetap, yang setara dengan memodelkan AR/MA/ARMA pada deret terdiferensi dan memudahkan peramalan pada skala asli. Syarat stasioneritas (akar AR) dan invertibilitas (akar MA) diperiksa lewat modulus akar terkecil yang harus > 1.

```python
def fit_arima(y, order, sorder=(0, 0, 0, 0)):
    # konstanta hanya dipakai bila tidak ada differencing sama sekali
    trend = 'c' if (order[1] + sorder[1]) == 0 else None
    try:
        mod = SARIMAX(y, order=order, seasonal_order=sorder, trend=trend,
                      enforce_stationarity=True, enforce_invertibility=True)
        return mod.fit(disp=False, maxiter=500)
    except Exception as e:
        print(f'   [gagal] {order}{sorder}: {e}')
        return None

def info_model(res):
    ar = np.abs(res.arroots) if len(res.arroots) else np.array([np.nan])
    ma = np.abs(res.maroots) if len(res.maroots) else np.array([np.nan])
    rv = getattr(res, 'mle_retvals', None) or {}
    return dict(k=len(res.params), AIC=res.aic, AICc=res.aicc, BIC=res.bic, logL=res.llf,
                akar_AR_min=np.nanmin(ar), akar_MA_min=np.nanmin(ma),
                Konvergen=bool(rv.get('converged', True)))

def lag_sig(nilai, n, lags):
    batas = 1.96 / np.sqrt(n)
    return [l for l in lags if l < len(nilai) and abs(nilai[l]) > batas]

def kand(sig):
    c = sorted(set(sig[:2]))
    if len(c) == 0: return [1, 2]
    if len(c) == 1: return [c[0], c[0] + 1]
    return c

def cocokkan(n, daftar, y=None):
    # daftar: list (order, sorder). Mengembalikan tabel dan dict model
    y = YT[n] if y is None else y
    baris, simpan = [], {}
    for od, so in daftar:
        res = fit_arima(y, od, so)
        if res is None:
            continue
        lab = f'ARIMA{od}' if so == (0, 0, 0, 0) else f'SARIMA{od}x{so}'
        inf = info_model(res)
        baris.append({'Model': lab, **inf})
        simpan[lab] = dict(res=res, order=od, sorder=so, d=od[1], D=so[1], label=lab, k_arma=od[0] + od[2] + so[0] + so[2])
    tab = pd.DataFrame(baris).set_index('Model') if baris else pd.DataFrame()
    return tab, simpan

def terbaik_aicc(tab):
    t = tab[tab['Konvergen']] if tab['Konvergen'].any() else tab
    return t['AICc'].idxmin()

# ---- ACF/PACF deret stasioner (diff biasa) ----
KAND = {}
fig, axes = plt.subplots(2, 2, figsize=(15, 8))
for i, n in enumerate(SERI):
    d = ORDE[n]['d_reg']
    w = beda(YT[n], d, 0)
    nn = len(w)
    L = min(24, nn // 2 - 1)
    plot_acf(w, lags=L, ax=axes[i, 0], title=f'ACF {n} stasioner (d={d})')
    plot_pacf(w, lags=L, ax=axes[i, 1], method='ywm', title=f'PACF {n} stasioner (d={d})')
    ac = acf(w, nlags=L, fft=False)
    pc = pacf(w, nlags=L, method='ywm')
    sig_p = lag_sig(pc, nn, range(1, 5))
    sig_q = lag_sig(ac, nn, range(1, 5))
    pk, qk = kand(sig_p), kand(sig_q)
    KAND[n] = dict(p=pk, q=qk, arma=list(dict.fromkeys([(pk[0], qk[0]), (1, 1), (pk[1], qk[1])])))
    print(f'{n}: lag PACF signifikan (1-4) = {sig_p} -> kandidat p = {pk};  lag ACF signifikan (1-4) = {sig_q} -> kandidat q = {qk};  kandidat ARMA = {KAND[n]["arma"]}')
plt.tight_layout()
plt.show()
```

```python
PIL = {n: {} for n in SERI}      # model terpilih per jenis (berdasarkan AICc data latih)
for n in SERI:
    d = ORDE[n]['d_reg']
    print('=' * 100)
    print(f'{n}: estimasi AR, MA, ARMA pada deret dengan d = {d} (data latih, skala model)')
    grup = {
        'AR':   [((p, d, 0), (0, 0, 0, 0)) for p in KAND[n]['p']],
        'MA':   [((0, d, q), (0, 0, 0, 0)) for q in KAND[n]['q']],
        'ARMA': [((p, d, q), (0, 0, 0, 0)) for (p, q) in KAND[n]['arma']],
    }
    for jenis, daftar in grup.items():
        tab, simpan = cocokkan(n, daftar)
        print(f'\n--- {jenis} ---  (akar_AR_min dan akar_MA_min harus > 1)')
        print(tab.round(3).to_string())
        lab = terbaik_aicc(tab)
        PIL[n][jenis] = simpan[lab]
        print(f'Terpilih ({jenis}): {lab}  [AICc data latih terkecil di antara kandidat dengan d yang sama]')
```

Koefisien yang tidak signifikan (kolom `P>|z|`) menandakan orde mungkin terlalu besar, tetapi **kecukupan model tidak ditentukan oleh p-value koefisien**; keputusan akhir bertumpu pada diagnostik residual (Tahap 7) dan akurasi data uji (Tahap 8). Karena AR/MA/ARMA di sini non-musiman, wajar bila residualnya masih menyisakan autokorelasi pada lag 12 untuk deret yang musimannya kuat. Hal itu yang ditangani SARIMA.

## **TAHAP 5. Model ARIMA(p, d, q)**

Kandidat: hasil identifikasi manual Tahap 4 (AR, MA, ARMA terbaik), ARIMA(1,d,1) sebagai pembanding umum, dan **satu hasil pencarian otomatis**: grid p, q = 0..3 pada *d* tetap dengan kriteria AICc (data latih), serta `pmdarima.auto_arima` bila terpasang. Seluruh kandidat memakai *d* dan data latih yang sama sehingga AIC/AICc/BIC **valid dibandingkan**.

```python
try:
    import pmdarima as pm
    ADA_PM = True
except Exception as e:
    ADA_PM = False
    print('pmdarima tidak tersedia; pencarian otomatis memakai grid AICc sendiri. Detail:', repr(e)[:120])

AUTO = {}
for n in SERI:
    d = ORDE[n]['d_reg']
    print('=' * 100)
    print(f'{n}: ARIMA dengan d = {d}')

    # (a) pencarian otomatis: grid p,q = 0..3
    grid = [((p, d, q), (0, 0, 0, 0)) for p in range(4) for q in range(4)]
    tab_g, simp_g = cocokkan(n, grid)
    lab_g = terbaik_aicc(tab_g)
    print('\nGrid otomatis (5 AICc terkecil):')
    print(tab_g.sort_values('AICc').head(5)[['k', 'AIC', 'AICc', 'BIC', 'Konvergen']].round(3).to_string())

    # (b) pmdarima (opsional): biarkan memilih d sendiri untuk dikritisi
    order_pm = None
    if ADA_PM:
        try:
            am = pm.auto_arima(YT[n].values, seasonal=False, d=None, start_p=0, start_q=0, max_p=4, max_q=4,
                               information_criterion='aicc', stepwise=True, suppress_warnings=True, error_action='ignore')
            order_pm = tuple(am.order)
            sama_d = order_pm[1] == d
            print(f'\npmdarima.auto_arima memilih ARIMA{order_pm}; d {"sama dengan" if sama_d else "BERBEDA dari"} d keputusan uji kita ({d}).')
            if not sama_d:
                print('   -> AIC tidak sebanding lintas d, sehingga model ini tidak dimasukkan ke perbandingan AICc.')
                order_pm = None
        except Exception as e:
            print('\npmdarima gagal dijalankan:', repr(e)[:150])

    # (c) kumpulkan kandidat unik
    ods = [PIL[n]['AR']['order'], PIL[n]['MA']['order'], PIL[n]['ARMA']['order'], (1, d, 1), simp_g[lab_g]['order']]
    if order_pm is not None:
        ods.append(order_pm)
    ods = list(dict.fromkeys(ods))
    tab_k, simp_k = cocokkan(n, [(o, (0, 0, 0, 0)) for o in ods])
    print('\nKandidat ARIMA yang dibandingkan (AICc/BIC valid: d dan data latih sama):')
    print(tab_k.round(3).to_string())
    lab = terbaik_aicc(tab_k)
    m = simp_k[lab]
    PIL[n]['ARIMA'] = m
    HASIL[n]['ARIMA'] = dict(kind='arima', nama=m['label'], res=m['res'], order=m['order'], sorder=m['sorder'],
                             d=m['d'], D=m['D'], k=m['k_arma'])
    AUTO[n] = simp_g[lab_g]
    print(f'\nARIMA terpilih: {lab}')

    # (d) kritik terhadap hasil otomatis
    r = np.asarray(AUTO[n]['res'].resid)[d:]
    lb = acorr_ljungbox(r, lags=[12], model_df=AUTO[n]['k_arma'], return_df=True)['lb_pvalue'].iloc[0]
    top2 = tab_g.sort_values('AICc')['AICc'].iloc[:2].values
    print('\nKritik hasil otomatis:')
    print(f'  - Pilihan grid {lab_g}; selisih AICc dengan peringkat kedua hanya {top2[1]-top2[0]:.2f} '
          + ('(< 2: kedua model praktis setara)' if top2[1] - top2[0] < 2 else '(cukup jelas)'))
    print(f'  - Ljung-Box lag 12 residual model otomatis: p = {lb:.4f} '
          + ('-> residual masih berautokorelasi; AICc kecil tidak menjamin model memadai' if lb < ALPHA else '-> tidak ada bukti autokorelasi pada lag 12'))
    print('  - Pencarian otomatis tidak mengenal musiman pada keluarga ARIMA non-musiman dan hanya mengoptimalkan AICc data latih.')
```

**Tafsir Tahap 5.** Model ARIMA terpilih adalah yang ber-AICc terkecil di antara kandidat dengan *d* dan data latih yang sama. Selisih AICc kurang dari sekitar 2 dianggap **tidak dapat dibedakan**; dalam kasus itu model yang lebih sederhana dan yang lulus diagnostik lebih diutamakan. Hasil otomatis tidak diterima begitu saja: ia hanya mengoptimalkan AICc data latih dan tidak memeriksa autokorelasi residual.

## **TAHAP 6. Model SARIMA(p, d, q)(P, D, Q)m**

Data bulanan dengan m = 12 dan lebih dari 3 siklus musiman pada data latih sehingga SARIMA dapat diestimasi (batas Tabel 1 terpenuhi). *d* dan *D* dari Tahap 2. Bagian musiman diidentifikasi dari lag 12 ACF/PACF deret yang sudah di-differencing: PACF signifikan di lag 12 menunjuk P = 1, ACF signifikan di lag 12 menunjuk Q = 1. Kandidat (P, Q) dari aturan tersebut digabung dengan (0,1), (1,0), (1,1), dan kandidat (p, q) non-musiman kecil; semuanya dibandingkan dengan **AICc data latih** pada (*d*, *D*, m) yang sama.

```python
SAR_TAB = {}
fig, axes = plt.subplots(2, 2, figsize=(15, 8))
for i, n in enumerate(SERI):
    d, D = ORDE[n]['d_sar'], ORDE[n]['D']
    z = beda(YT[n], d, D)
    nn = len(z)
    L = min(36, nn // 2 - 1)
    plot_acf(z, lags=L, ax=axes[i, 0], title=f'ACF {n} (d={d}, D={D})')
    plot_pacf(z, lags=L, ax=axes[i, 1], method='ywm', title=f'PACF {n} (d={d}, D={D})')
    ac = acf(z, nlags=L, fft=False)
    pc = pacf(z, nlags=L, method='ywm')
    b = 1.96 / np.sqrt(nn)
    P_r = int(len(pc) > M and abs(pc[M]) > b)
    Q_r = int(len(ac) > M and abs(ac[M]) > b)
    sig_p = lag_sig(pc, nn, range(1, 5))
    sig_q = lag_sig(ac, nn, range(1, 5))
    print(f'{n}: lag-12 PACF = {pc[M] if len(pc) > M else np.nan:.2f}, ACF = {ac[M] if len(ac) > M else np.nan:.2f} (batas +/-{b:.2f}) -> aturan: P={P_r}, Q={Q_r}; '
          f'lag non-musiman signifikan PACF {sig_p}, ACF {sig_q}')
    pq = {(0, 0), (1, 0), (0, 1), (1, 1)}
    if sig_p or sig_q:
        pq.add((min(sig_p[:1] or [0]), min(sig_q[:1] or [0])))
    PQ = {(0, 1), (1, 0), (1, 1)}
    if (P_r, Q_r) != (0, 0):
        PQ.add((P_r, Q_r))
    SAR_TAB[n] = dict(d=d, D=D, pq=sorted(pq), PQ=sorted(PQ))
plt.tight_layout()
plt.show()
```

```python
for n in SERI:
    d, D = SAR_TAB[n]['d'], SAR_TAB[n]['D']
    daftar = [((p, d, q), (P, D, Q, M)) for (p, q) in SAR_TAB[n]['pq'] for (P, Q) in SAR_TAB[n]['PQ']]
    print('=' * 100)
    print(f'{n}: {len(daftar)} kandidat SARIMA dengan d={d}, D={D}, m={M} (data latih). Proses ini dapat memakan waktu beberapa menit.')
    tab, simpan = cocokkan(n, daftar)
    if tab.empty:
        print('Tidak ada model SARIMA yang berhasil diestimasi. Laporkan alasan ketidakstabilan estimasi.')
        continue
    print(tab.sort_values('AICc').head(8)[['k', 'AIC', 'AICc', 'BIC', 'akar_AR_min', 'akar_MA_min', 'Konvergen']].round(3).to_string())
    n_gagal = int((~tab['Konvergen']).sum())
    print(f'\nModel tidak konvergen: {n_gagal} dari {len(tab)}' + (' (dikeluarkan dari pemilihan bila ada alternatif konvergen)' if n_gagal else ''))
    lab = terbaik_aicc(tab)
    m = simpan[lab]
    PIL[n]['SARIMA'] = m
    HASIL[n]['SARIMA'] = dict(kind='arima', nama=m['label'], res=m['res'], order=m['order'], sorder=m['sorder'],
                              d=m['d'], D=m['D'], k=m['k_arma'])
    print(f'SARIMA terpilih: {lab}')
    print(m['res'].summary().tables[1])
```

**Tafsir Tahap 6.** Bila beberapa kandidat SARIMA berselisih AICc kurang dari sekitar 2, pilihan di antaranya tidak dapat dibedakan secara statistik, dan model sederhana lebih diutamakan. Estimasi SARIMA pada data latih yang pendek rapuh (lihat catatan Tabel 1 instruksi); jika ada model yang tidak konvergen atau koefisien musiman mendekati batas stasioneritas/invertibilitas (akar mendekati 1), laporkan hal itu sebagai keterbatasan.

## **TAHAP 7. Diagnostik residual**

Untuk setiap model akhir (SES, Holt, Holt-Winters, AR, MA, ARMA, ARIMA, SARIMA) diperiksa: plot residual, ACF residual, uji **Ljung-Box** (H0: residual white noise; derajat bebas dikoreksi dengan jumlah parameter ARMA), normalitas (**Jarque-Bera** dan Shapiro-Wilk, plus QQ-plot), kestabilan varians (uji Levene paruh awal vs paruh akhir dan uji ARCH). Residual awal yang hanya artefak inisialisasi differencing dibuang.

```python
MODEL_URUT = ['SES', 'Holt', 'Holt-Winters', 'AR', 'MA', 'ARMA', 'ARIMA', 'SARIMA']

def resid_model(info):
    if info['kind'] == 'es':
        return np.asarray(info['fit'].resid, dtype=float)
    burn = info['d'] + info['D'] * M
    return np.asarray(info['res'].resid, dtype=float)[burn:]

def diagnosa(resid, k):
    r = pd.Series(np.asarray(resid, dtype=float)).dropna().reset_index(drop=True)
    n = len(r)
    lags = [l for l in (6, 12, 24) if l > k and l < n // 2]
    if not lags:
        lags = [max(k + 1, 2)]
    lb = acorr_ljungbox(r, lags=lags, model_df=k, return_df=True)
    jb = jarque_bera(r)
    sh = stats.shapiro(r)
    lev = stats.levene(r[: n // 2], r[n // 2:])
    try:
        arch_p = het_arch(r, nlags=min(6, max(1, n // 6)))[1]
    except Exception:
        arch_p = np.nan
    return dict(n_resid=n, LB_p_min=float(lb['lb_pvalue'].min()), LB_lags=lags, JB_p=float(jb[1]), Shapiro_p=float(sh[1]),
                Levene_p=float(lev[1]), ARCH_p=float(arch_p), skew=float(jb[2]), kurt=float(jb[3]))

DIAG = {n: {} for n in SERI}
for n in SERI:
    for nm in MODEL_URUT:
        if nm not in HASIL[n]:
            continue
        info = HASIL[n][nm]
        d_ = diagnosa(resid_model(info), info['k'])
        d_['lulus_LB'] = d_['LB_p_min'] >= ALPHA
        d_['normal'] = d_['JB_p'] >= ALPHA and d_['Shapiro_p'] >= ALPHA
        d_['var_stabil'] = d_['Levene_p'] >= ALPHA and (np.isnan(d_['ARCH_p']) or d_['ARCH_p'] >= ALPHA)
        DIAG[n][nm] = d_

    tab = pd.DataFrame.from_dict(DIAG[n], orient='index')[['n_resid', 'LB_p_min', 'JB_p', 'Shapiro_p', 'Levene_p', 'ARCH_p', 'skew', 'kurt',
                                    'lulus_LB', 'normal', 'var_stabil']]
    tab.insert(0, 'Spesifikasi', [HASIL[n][m]['nama'] for m in tab.index])
    print('=' * 110)
    print(f'{n}: ringkasan diagnostik residual (p-value; lulus bila p >= {ALPHA})')
    print(tab.to_string(float_format=lambda v: f'{v:.4f}'))

    print('\nAkibat pada interval peramalan:')
    for nm in tab.index:
        dd = DIAG[n][nm]
        akibat = []
        if not dd['lulus_LB']:
            akibat.append('masih ada autokorelasi sisa (struktur belum tertangkap; galat ramalan berkorelasi, interval prediksi cenderung terlalu sempit)')
        if not dd['normal']:
            akibat.append('residual tidak normal (interval berbasis normal kurang akurat, terutama pada ekor)')
        if not dd['var_stabil']:
            akibat.append('varians tidak stabil (lebar interval konstan tidak mencerminkan ketidakpastian yang berubah)')
        print(f'  {nm:13s}: ' + ('lulus semua uji' if not akibat else '; '.join(akibat)))
```

```python
for n in SERI:
    fig, axes = plt.subplots(len(MODEL_URUT), 3, figsize=(16, 2.7 * len(MODEL_URUT)))
    for i, nm in enumerate(MODEL_URUT):
        if nm not in HASIL[n]:
            continue
        r = resid_model(HASIL[n][nm])
        Lr = max(2, min(24, len(r) // 2 - 1))
        axes[i, 0].plot(r, color='tab:blue', lw=1)
        axes[i, 0].axhline(0, color='k', lw=0.8)
        axes[i, 0].set_title(f'{n} - {nm}: residual', fontsize=10)
        plot_acf(r, lags=Lr, ax=axes[i, 1], title=f'{n} - {nm}: ACF residual')
        stats.probplot(r, dist='norm', plot=axes[i, 2])
        axes[i, 2].set_title(f'{n} - {nm}: QQ-plot', fontsize=10)
    plt.tight_layout()
    plt.show()
```

**Tafsir Tahap 7.** Model yang **lulus** adalah yang p-value Ljung-Box-nya >= 0,05 pada semua lag yang diuji (tidak ada bukti autokorelasi sisa). Model yang gagal Ljung-Box tidak boleh dinyatakan "memenuhi asumsi"; interval prediksinya perlu dibaca sebagai terlalu optimistis. Ketidaknormalan residual (mis. karena pencilan atau lonjakan 2025) membuat interval 80%/95% berbasis normal hanya pendekatan. Exponential Smoothing diperiksa dengan cara yang sama, tetapi uji Ljung-Box pada residual satu-langkah ES hanya indikatif.

## **TAHAP 8. Pembandingan model**

Semua model akhir meramal 12 bulan data uji dari **model yang diestimasi pada data latih saja** (tanpa pembaruan), lalu dievaluasi pada skala asli (ton) dengan RMSE, MAE, dan MAPE. MAPE aman dipakai karena seluruh nilai deret jauh dari nol. Dua pembanding sederhana ditambahkan sebagai tolok ukur: **naif musiman** (nilai bulan yang sama tahun lalu) dan **MA(12)** (rata-rata 12 bulan terakhir data latih).

AIC/BIC ditampilkan hanya untuk keluarga ARIMA dan **hanya sah dibandingkan di dalam grup (d, D) yang sama**; ES tidak memiliki AIC yang sebanding dengan ARIMA. Pemilihan akhir memakai RMSE data uji, dengan aturan ketidakpastian di bawah.

```python
def metrik(a, f):
    a = np.asarray(a, dtype=float); f = np.asarray(f, dtype=float)
    e = a - f
    return dict(RMSE=float(np.sqrt(np.mean(e ** 2))), MAE=float(np.mean(np.abs(e))), MAPE=float(np.mean(np.abs(e / a)) * 100))

def ramal_uji(n, info):
    if info['kind'] == 'es':
        f = np.asarray(info['fit'].forecast(H_UJI), dtype=float)
    else:
        f = inv(n, np.asarray(info['res'].get_forecast(H_UJI).predicted_mean, dtype=float))
    return pd.Series(f, index=TEST[n].index)

FC, TAB_BAND, PILIH = {}, {}, {}
for n in SERI:
    FC[n] = {}
    for nm in MODEL_URUT:
        if nm in HASIL[n]:
            FC[n][nm] = ramal_uji(n, HASIL[n][nm])
    FC[n]['Naif musiman'] = pd.Series(TRAIN[n].iloc[-M:].values, index=TEST[n].index)
    FC[n]['MA(12)'] = pd.Series(np.repeat(TRAIN[n].iloc[-M:].mean(), H_UJI), index=TEST[n].index)

    baris = []
    for nm, f in FC[n].items():
        mt = metrik(TEST[n].values, f.values)
        row = {'Model': nm, 'Spesifikasi': HASIL[n][nm]['nama'] if nm in HASIL[n] else '(pembanding)'}
        if nm in HASIL[n] and HASIL[n][nm]['kind'] == 'arima':
            res = HASIL[n][nm]['res']
            row.update({'Grup (d,D)': f'({HASIL[n][nm]["d"]},{HASIL[n][nm]["D"]})', 'AIC': res.aic, 'BIC': res.bic})
        else:
            row.update({'Grup (d,D)': '-', 'AIC': np.nan, 'BIC': np.nan})
        row.update(mt)
        row['Lulus Ljung-Box'] = DIAG[n][nm]['lulus_LB'] if nm in DIAG[n] else np.nan
        row['k'] = HASIL[n][nm]['k'] if nm in HASIL[n] else 0
        baris.append(row)
    tab = pd.DataFrame(baris).set_index('Model')
    TAB_BAND[n] = tab
    print('=' * 120)
    print(f'{n}: PERBANDINGAN MODEL (metrik pada data uji, ton; AIC/BIC hanya dibandingkan dalam grup (d,D) sama)')
    print(tab.sort_values('RMSE').to_string(float_format=lambda v: f'{v:,.3f}' if abs(v) < 1000 else f'{v:,.0f}'))

    # ---- aturan pemilihan ----
    kandidat = tab.loc[[m for m in MODEL_URUT if m in tab.index]]
    rmse_min = kandidat['RMSE'].min()
    dekat = kandidat[kandidat['RMSE'] <= 1.05 * rmse_min]
    lulus = dekat[dekat['Lulus Ljung-Box'] == True]
    pool = lulus if len(lulus) else dekat
    terpilih = pool.sort_values(['k', 'RMSE']).index[0]
    PILIH[n] = terpilih
    print(f'\nModel dengan RMSE terkecil     : {kandidat["RMSE"].idxmin()} (RMSE = {rmse_min:,.0f} ton)')
    if len(dekat) > 1:
        print(f'Model dengan selisih RMSE <= 5% : {list(dekat.index)} -> TIDAK DAPAT DIBEDAKAN secara praktis (hanya {H_UJI} titik uji).')
        print('   Aturan penentu: utamakan yang lulus Ljung-Box, lalu yang paling sedikit parameter.')
    print(f'MODEL TERPILIH {n}: {terpilih} ({HASIL[n][terpilih]["nama"]})')
    for b in ('Naif musiman', 'MA(12)'):
        if tab.loc[terpilih, 'RMSE'] >= tab.loc[b, 'RMSE']:
            print(f'   PERINGATAN: model terpilih tidak lebih baik daripada pembanding {b} (RMSE {tab.loc[b,"RMSE"]:,.0f}). Kompleksitas model tidak terbukti bermanfaat.')
    print()
```

## **TAHAP 9. Peramalan**

Model terpilih **diestimasi ulang pada seluruh data** (latih + uji) dengan spesifikasi/orde yang sama, lalu meramal **12 bulan ke depan** (September 2026 sampai Agustus 2027) dengan interval prediksi 80% dan 95%. Pada model ARIMA/SARIMA interval berasal dari varians prediksi model pada skala model lalu ditransformasi balik (sehingga asimetris bila memakai log). Pada Exponential Smoothing interval berasal dari simulasi galat (seed tetap).

```python
def ramal_depan(n, nama_model, h=H_RAMAL):
    info = HASIL[n][nama_model]
    full = SERI[n]
    idx = pd.date_range(full.index[-1] + pd.offsets.MonthBegin(1), periods=h, freq='MS')
    if info['kind'] == 'arima':
        res = fit_arima(fwd(n, full), info['order'], info['sorder'])
        fc = res.get_forecast(h)
        mean = inv(n, np.asarray(fc.predicted_mean, dtype=float))
        c80 = np.asarray(fc.conf_int(alpha=0.20)); c95 = np.asarray(fc.conf_int(alpha=0.05))
        out = pd.DataFrame({'Ramalan': mean, 'Bawah80': inv(n, c80[:, 0]), 'Atas80': inv(n, c80[:, 1]),
                            'Bawah95': inv(n, c95[:, 0]), 'Atas95': inv(n, c95[:, 1])}, index=idx)
        return out, 'interval analitik model (state-space)'
    f = fit_es(full, info['spek'])
    mean = np.asarray(f.forecast(h), dtype=float)
    try:
        sim = f.simulate(h, repetitions=2000, error='mul' if info['spek']['seasonal'] == 'mul' else 'add', random_state=SEED)
        sim = np.asarray(sim, dtype=float)
        q = lambda p: np.quantile(sim, p, axis=1)
        metode = 'interval simulasi galat (2000 lintasan, seed tetap)'
        out = pd.DataFrame({'Ramalan': mean, 'Bawah80': q(0.10), 'Atas80': q(0.90), 'Bawah95': q(0.025), 'Atas95': q(0.975)}, index=idx)
    except Exception as e:
        s = np.nanstd(np.asarray(f.resid, dtype=float))
        metode = f'interval pendekatan normal dengan SD residual tetap (simulasi gagal: {repr(e)[:60]})'
        out = pd.DataFrame({'Ramalan': mean, 'Bawah80': mean - 1.2816 * s, 'Atas80': mean + 1.2816 * s,
                            'Bawah95': mean - 1.96 * s, 'Atas95': mean + 1.96 * s}, index=idx)
    return out, metode

RAMALAN = {}
for n in SERI:
    nm = PILIH[n]
    out, metode = ramal_depan(n, nm)
    RAMALAN[n] = out
    out.round(0).to_csv(f'ramalan_{n.lower()}_12bulan.csv')
    print('=' * 100)
    print(f'{n}: peramalan 12 bulan ke depan dengan {nm} ({HASIL[n][nm]["nama"]}); {metode}')
    print(out.round(0).to_string(float_format=lambda v: f'{v:,.0f}'))

    hist = SERI[n].iloc[-12:]
    total_f, total_h = out['Ramalan'].sum(), hist.sum()
    print(f'\nTotal ramalan 12 bulan = {total_f:,.0f} ton; total 12 bulan terakhir aktual = {total_h:,.0f} ton ({(total_f/total_h-1)*100:+.1f}%).')
    print(f'Bulan puncak ramalan: {out["Ramalan"].idxmax():%B %Y} ({out["Ramalan"].max():,.0f} ton); bulan terendah: {out["Ramalan"].idxmin():%B %Y} ({out["Ramalan"].min():,.0f} ton).')
    lebar1 = (out['Atas95'].iloc[0] - out['Bawah95'].iloc[0]) / out['Ramalan'].iloc[0]
    lebarh = (out['Atas95'].iloc[-1] - out['Bawah95'].iloc[-1]) / out['Ramalan'].iloc[-1]
    print(f'Lebar interval 95% relatif terhadap ramalan: {lebar1*100:.0f}% (bulan ke-1) dan {lebarh*100:.0f}% (bulan ke-{H_RAMAL}).')
    if nm in DIAG[n] and not DIAG[n][nm]['lulus_LB']:
        print('CATATAN: model terpilih tidak lulus Ljung-Box; interval prediksi kemungkinan terlalu sempit.')
    if nm in DIAG[n] and not DIAG[n][nm]['normal']:
        print('CATATAN: residual model terpilih tidak normal; interval 80%/95% hanya pendekatan.')

fig, axes = plt.subplots(2, 1, figsize=(14, 11))
for ax, n in zip(axes, SERI):
    out = RAMALAN[n]
    hist = SERI[n].iloc[-48:]
    ax.plot(hist, color='black', marker='o', ms=3, label='Aktual')
    ax.plot(out['Ramalan'], color='tab:red', marker='o', ms=3, label=f'Ramalan ({PILIH[n]})')
    ax.fill_between(out.index, out['Bawah95'], out['Atas95'], color='tab:red', alpha=0.15, label='Interval 95%')
    ax.fill_between(out.index, out['Bawah80'], out['Atas80'], color='tab:red', alpha=0.30, label='Interval 80%')
    ax.set_title(f'Peramalan Produksi {n} 12 Bulan ke Depan (model estimasi ulang pada seluruh data)', fontweight='bold')
    ax.set_xlabel('Tanggal'); ax.set_ylabel('Produksi (ton)')
    ax.grid(True, alpha=0.4); ax.legend()
plt.tight_layout()
plt.show()
```
